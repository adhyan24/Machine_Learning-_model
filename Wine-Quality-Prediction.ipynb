## Loading Dataset


```python
dataset= pd.read_csv("winequality-red.csv", sep=";")
print(dataset.shape)  # (1599, 12)
```

```python
# numbers of values for each qality
sns.catplot(x= 'quality' , data = dataset , kind = 'count' , color = 'black')
```

```python
# volatile acidity  vs quality
plot  = plt.figure(figsize = (5,5))
sns.barplot(x='quality' , y = 'volatile acidity' , data = dataset , color = 'red')
```

```python
# citric acid  vs quality
plot  = plt.figure(figsize = (5,5))
sns.barplot(x='quality' , y = 'citric acid' , data = dataset , color = 'yellow')
```

```python
# chlorides vs quality
plot  = plt.figure(figsize = (5,5))
sns.barplot(x='quality' , y = 'chlorides' , data = dataset , color = 'green')
```
